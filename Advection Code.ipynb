```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, ArtistAnimation
from IPython.display import HTML
```

## Finite-difference Method

### Forward difference

u(i,t+1) = u(i,t) - c*dt/dx*( u(i+1,t) - u(i,t) )

```python
def square_wave(x):
    if 0.25 <= x <= 0.75:
        return 1.0
    else:
        return 0.0
    
def forward_differece(u, c, dt, dx):
    u_old = u.copy()
    for i in range(1, len(u)-1):
        u[i] = u_old[i] - c* dt/dx * (u_old[i+1] - u_old[i])
    return u
```

```python
ngrids = 50
L = 1.0
dx = L / ngrids
c =1
cfl = 0.8
dt = cfl * dx / c
u = np.array([square_wave(x) for x in np.linspace(0, L, ngrids)])
x = np.linspace(0, L, ngrids)
u_old = u.copy()
u_sol = np.zeros(ngrids)
error = np.zeros(ngrids)
error_fwd = []

fig, (ax1,ax2) = plt.subplots(2,1)
line_num, = ax1.plot(x, u, label="result", color="r")
line_exact, = ax1.plot(x, u_sol, label="Exact", color="b")
line_error, = ax2.plot(x, error, label="Error", color="g")
ax1.set_xlim(0, L)
ax1.legend(loc='upper right')
ax2.set_ylim(-2, 2)
fig.supxlabel(r'$x$')
fig.supylabel(r'$u$')


def update(frame):
    forward_differece(u, c, dt, dx)
    for i in range(len(x)-1):
        u_sol[i] = square_wave(x[i]- c*frame*dt)
    error = u_sol - u
    error_fwd.append(np.linalg.norm(error))
    line_error.set_ydata(error)
    line_num.set_ydata(u)
    line_exact.set_ydata(u_sol)
    return line_num, line_exact, line_error


plt.close(fig)
ani = FuncAnimation(fig, update,frames = 100, interval = 100)
HTML(ani.to_jshtml())
display(HTML(ani.to_jshtml()))
print(np.sqrt(np.sum(np.square(error_fwd))/ngrids))
```

### Backward difference

u(i,t+1) = u(i,t) - c*dt/dx*( u(i,t) - u(i-1,t) )

```python
def square_wave(x):
    if 0.25 <= x <= 0.75:
        return 1.0
    else:
        return 0.0
    
def backward_differece(u, c, dt, dx):
    u_old = u.copy()
    for i in range(1, len(u)-1):
            u[i] = u_old[i] - c* dt/dx * (u_old[i] - u_old[i-1])
    return u
```

```python
ngrids = 100
L = 1.0
dx = L / ngrids
c =1
cfl = 0.8
dt = cfl * dx / c
u = np.array([square_wave(x) for x in np.linspace(0, L, ngrids)])
x = np.linspace(0, L, ngrids)
u_sol = np.zeros(ngrids)
error = np.zeros(ngrids)
error_bwd = []
fig, (ax1,ax2) = plt.subplots(2,1)
line_num, = ax1.plot(x, u, label="Numerical", color="r")
line_exact, = ax1.plot(x, u_sol, label="Exact", color="b")
line_error, = ax2.plot(x, error, label="Error", color="g")


ax1.legend(loc='upper right')
fig.tight_layout()
ax2.legend(loc='upper right')
ax2.set_ylim(-2, 2)
fig.supxlabel(r'$x$')
fig.supylabel(r'$u$')

def back_update(frame):
    backward_differece(u, c, dt, dx)
    u_sol = np.array([square_wave(xi- c*frame*dt) for xi in x])
    error = u_sol - u
    error_bwd.append(np.linalg.norm(error))
    line_num.set_ydata(u)
    line_exact.set_ydata(u_sol)
    line_error.set_ydata(error)
    return line_num, line_exact, line_error



plt.close(fig)
ani_back = FuncAnimation(fig,back_update, frames=100, interval =100)
HTML(ani_back.to_jshtml())
display(HTML(ani_back.to_jshtml()))
print(np.sqrt(np.sum(np.square(error_bwd))/ngrids))

```

### Central difference

u(i,t+1) = u(i,t) - c*dt/dx*( u(i+1,t) - u(i-1,t) )/2

```python
def square_wave(x):
    if 0.25 <= x <= 0.75:
        return 1.0
    else:
        return 0.0
    
def central_differece(u, c, dt, dx):
    u_old = u.copy()
    for i in range(1, len(u)-1):
        u[i] = u_old[i] - c* dt/dx * (u_old[i+1] - u_old[i-1])/2
    return u            
```

```python
ngrids = 100
L = 1.0
dx = L / ngrids
c =1
cfl = 0.8
dt = cfl * dx / c
u = np.array([square_wave(x) for x in np.linspace(0, L, ngrids)])
x = np.linspace(0, L, ngrids)
u_sol = np.zeros(ngrids)
error = np.zeros(ngrids)
error_cen = []


fig,(ax1,ax2) = plt.subplots(2,1)
plt.close(fig)
fig.tight_layout()
ax2.set_ylim(-2, 2)
fig.supxlabel(r'$x$')
fig.supylabel(r'$u$')

line_num, = ax1.plot(x,u,label="numerical",color="r")
line_exact, = ax1.plot(x,u_sol,label="exact", color="b")
line_error, = ax2.plot(x,error,label="error", color="g")

ax1.legend(loc='upper right')
ax2.legend(loc='upper right')


def cen_update(frame):
    central_differece(u, c, dt, dx)
    u_sol = np.array([square_wave(xi- c*frame*dt) for xi in x])
    error = u_sol - u
    error_cen.append(np.linalg.norm(error))
    line_num.set_ydata(u)
    line_exact.set_ydata(u_sol)
    line_error.set_ydata(error)
    return line_num, line_exact, line_error

ani_cen = FuncAnimation(fig,cen_update, frames=100, interval =100)
HTML(ani_cen.to_jshtml())
display(HTML(ani_cen.to_jshtml()))
print(np.sqrt(np.sum(np.square(error_cen))/ngrids))
```

## Finite- Volume Method

del(u)/dt + del(f)/dt = 0 

f(j,n+1) = 0.5*(c*u(i,n) + c*u(i+1,n) - abs(c)*(u(n,i+1) - u(i,n)))

u(i,n+1) = u(i,n) - dt/dx*(f(j)-f(j-1))

```python
ngrids = 30
L = 1.0
dx = L / ngrids
c = 1
cfl = 0.8
dt = cfl * dx/c
artistA = []
ghost_cells = 1
ntot = ngrids + 2*ghost_cells
ist = 1
ien = ist + ngrids



def square_wave(x):
    if abs(x-0.5) < 0.1:
        return 1.0
    else:
        return 0.0

def animation(x ,u,time):

    im1 = ax.plot(x, u,'o-', color = "r")
    for i in range(ntot):
        x_sol = x[i] - c*time
        while x_sol < 0:
            x_sol = x_sol + L
        u_sol[i] = square_wave(x_sol)
        
    im2 = ax.plot(x, u_sol,'-', color = "b")
    fig.tight_layout()
    artistA.append(im1 + im2)

time = 0
t_out = 1
u = np.zeros(ntot)
f = np.zeros(ntot)
u_sol = np.zeros(ntot)

cel_center = np.zeros(ntot)
cel_boundary = np.zeros(ntot+1)

fig, ax = plt.subplots()
plt.xlim(0,L)
plt.ylim(-0.5,1.5)
plt.close(fig)
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$u$') 

for i in range(ntot +1):
        cel_boundary[i] = (i - ist) * dx

for i in range(ntot):
        cel_center[i] = 0.5*(cel_boundary[i] + cel_boundary[i+1])

for i in range(ntot):
    u[i] = square_wave(cel_center[i])



animation(cel_center, u, time)

while time < t_out:

    for i in range(ist-1, ien):
        f[i] = 0.5*c*(u[i] + u[i+1]) - 0.5*abs(c)*(u[i+1] - u[i])

    for i in range(ist,ien):        
        u[i] = u[i] - dt/dx * (f[i] - f[i-1])
        
    time = time + dt
    animation(cel_center, u, time)

ax.legend(['Numerical', 'Exact'], loc='upper right')
ani_vol = ArtistAnimation(fig, artistA, interval=100)
HTML(ani_vol.to_jshtml())
```
